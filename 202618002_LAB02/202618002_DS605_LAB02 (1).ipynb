TASK 2:

● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @
or np.matmul().

● For a square matrix, find its transpose and determinant.

● If invertible, find the inverse and verify A @ A^-1 using np.allclose().

● Use vectorized NumPy operations; avoid explicit Python loops.

```python
#2x2 matrices
matrix_A = np.array([
    [3, 12, 9],
    [4, 6, 14],
    [5, 10, 11]
])

matrix_B = np.array([
    [6, 19, 23],
    [7, 4, 16],
    [3, 4, 5]
])

print("Matrix A:\n", matrix_A)
print("\nMatrix B:\n", matrix_B)


# 2. Addition
addition_result = matrix_A + matrix_B

print("\nAddition (A + B):\n", addition_result)


# 3. Element-wise Multiplication
element_wise_result = matrix_A * matrix_B

print("\nElement-wise Multiplication (A * B):\n", element_wise_result)


# 4. Matrix Multiplication
matrix_mult_result = matrix_A @ matrix_B


print("\nMatrix Multiplication (A @ B):\n", matrix_mult_result)

# 5. Matrix transpose
A_t = matrix_A.T
print("Transpose of A:\n", A_t)

# 6. Determinant
det_A = np.linalg.det(matrix_A)
print(f"Determinant of A: {det_A:.2f}")
```

```python
if abs(det_A)!=0:
    print("Matrix is invertible.")
    A_inverse = np.linalg.inv(matrix_A)
    print("Inverse:\n", np.round(A_inverse, 2))
else:
    print("Matrix is singular.")
```

loc and iloc are used in pandas to select specific rows and columns from a DataFrame. loc performs label-based indexing, while iloc performs position-based indexing. For example, here, df.loc[4, ["Name","Sex"]] selects the Name and sex columns of the row with index label 4. On the other hand, df.iloc[4, [3,4]] selects the value at the third row and third and fourth column based on their integer positions.

```python
# Fill missing Age with mean Age; show before/after
print("Missing Age BEFORE:", df["Age"].isnull().sum())

df_mean_filled = df.copy()
df_mean_filled["Age"] = df_mean_filled["Age"].fillna(df_mean_filled["Age"].mean())

print("Missing Age AFTER (mean fill):", df_mean_filled["Age"].isnull().sum())
```

```python
rng = np.random.default_rng(42)
# imputing with mean, median, mode, and a random value drawn from existing Age values
age_mean = df["Age"].mean()
age_median = df["Age"].median()
age_mode = df["Age"].mode()[0]

df_impute = df.copy()
df_impute["Age_mean_imputed"] = df["Age"].fillna(age_mean)
df_impute["Age_median_imputed"] = df["Age"].fillna(age_median)
df_impute["Age_mode_imputed"] = df["Age"].fillna(age_mode)

# Random-value imputation: fill NaNs with random draws from the observed (non-null) Age values
observed_ages = df["Age"].dropna().values
n_missing = df["Age"].isnull().sum()
random_fill_values = rng.choice(observed_ages, size=n_missing, replace=True)

df_impute["Age_random_imputed"] = df["Age"].copy()
missing_mask = df_impute["Age_random_imputed"].isnull()
df_impute.loc[missing_mask, "Age_random_imputed"] = random_fill_values

print(df_impute[["Age", "Age_mean_imputed", "Age_median_imputed",
                  "Age_mode_imputed", "Age_random_imputed"]].head(10))
```

```python
#for fare, printing q1,q3,IQR,1.5xIQR,outlier count
print("Q1:",df.Fare.quantile(0.25))
print("Q3:",df.Fare.quantile(0.75))
print("IQR:",df.Fare.quantile(0.75)-df.Fare.quantile(0.25))
print("1.5 IQR:",1.5*(df.Fare.quantile(0.75)-df.Fare.quantile(0.25)))
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# outliers
# We create a boolean mask (True/False) for fares outside the bounds, then sum the True values
outliers_mask = (df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)
outliers_count = outliers_mask.sum()

print("--- Fare Outlier Statistics ---")
print(f"Q1 (25th Percentile): {Q1:.4f}")
print(f"Q3 (75th Percentile): {Q3:.4f}")
print(f"IQR:                  {IQR:.4f}")
print("\n\n")
print(f"Lower Bound:          {lower_bound:.4f}")
print(f"Upper Bound:          {upper_bound:.4f}")
print("\n\n")
print(f"Total Outliers Found: {outliers_count}")
```
